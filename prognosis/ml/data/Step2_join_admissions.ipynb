```python
import pandas as pd
import re
import os
import re
```

```python
DATA_DIR_ICD10 = '../data/'
DATA_DIR = '../../data/mimic-4/'
```

```python
df = pd.read_csv(os.path.join(DATA_DIR_ICD10, "diagnoses_icd10.csv"), index_col=0)
admissions = pd.read_csv(os.path.join(DATA_DIR, "admissions.csv"))
df = pd.merge(
    df,
    admissions[['hadm_id', 'admittime', 'dischtime', 'deathtime']],
    on='hadm_id'
)
```

```python
df = df.sort_values(["subject_id", "admittime", "hadm_id", "seq_num"])
df_clean = df.drop_duplicates(subset=["subject_id", "hadm_id", "icd10_category"], keep="first").copy()
```
